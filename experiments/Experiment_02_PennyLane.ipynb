## Aim

To implement an Ry gate with rotation angle π/3 and observe the resulting quantum state.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a one-qubit PennyLane device and QNode.
2. Apply Ry(π/3) to the qubit.
3. Measure the state and display the circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)

@qml.qnode(dev)
def circuit():
    qml.RY(np.pi/3,wires=0)
    return qml.state()

state = circuit()

print("Quantum state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The Ry(π/3) gate was successfully implemented. The resulting state vector shows the expected Y-axis rotation of the qubit.
