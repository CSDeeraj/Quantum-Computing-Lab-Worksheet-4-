## Aim

To construct a two-qubit circuit by applying Rx to the first qubit and Ry to the second qubit.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a two-qubit PennyLane device and QNode.
2. Apply Rx(π/4) to qubit 0 and Ry(π/3) to qubit 1.
3. Display the resulting state and circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=2)

@qml.qnode(dev)
def circuit():
    qml.RX(np.pi/4,wires=0)
    qml.RY(np.pi/3,wires=1)
    return qml.state()

state = circuit()

print("Two-qubit quantum state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The two-qubit circuit was successfully simulated. Rx was applied to the first qubit and Ry to the second qubit, producing the combined two-qubit state.
