## Aim

To apply a Hadamard gate followed by Rx rotations with angles π/6, π/4, and π/3 and observe the final states.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a one-qubit PennyLane device and QNode.
2. Apply a Hadamard gate followed by Rx for each specified angle.
3. Display the final state and circuit for each angle.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)
angles = [np.pi/6,np.pi/4,np.pi/3]

for theta in angles:
    @qml.qnode(dev)
    def circuit():
        qml.Hadamard(wires=0)
        qml.RX(theta,wires=0)
        return qml.state()

    print(f"\nRx angle = {theta:.6f} radians")
    print("Final state:",circuit())
    print("Circuit:")
    print(qml.draw(circuit)())
```

## Result

The Hadamard gate first creates a superposition, and the subsequent Rx rotation changes the state according to the selected angle. Different angles produce different final states.
