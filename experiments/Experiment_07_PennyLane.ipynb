## Aim

To vary the Ry rotation angle from 0 to 2π and study the expectation value of the Pauli-Z operator.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy
- Matplotlib


## Procedure

1. Create a one-qubit PennyLane device and QNode.
2. Evaluate the Pauli-Z expectation value for angles from 0 to 2π.
3. Plot expectation value against the rotation angle.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np
import matplotlib.pyplot as plt

dev = qml.device("default.qubit",wires=1)

@qml.qnode(dev)
def circuit(theta):
    qml.RY(theta,wires=0)
    return qml.expval(qml.PauliZ(0))

angles = np.linspace(0,2*np.pi,100)
values = np.array([circuit(theta) for theta in angles])

print("Expectation value at θ = 0:",circuit(0))
print("Expectation value at θ = π:",circuit(np.pi))

plt.plot(angles,values)
plt.xlabel("Rotation angle θ")
plt.ylabel("<Z>")
plt.title("Pauli-Z Expectation Value vs Ry Rotation Angle")
plt.grid(True)
plt.show()
```

## Result

The Pauli-Z expectation value varies periodically as the Ry angle changes from 0 to 2π, demonstrating the continuous evolution of the qubit state.
