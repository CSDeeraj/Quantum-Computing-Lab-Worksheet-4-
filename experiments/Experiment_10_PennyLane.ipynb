## Aim

To construct a two-qubit circuit using Ry, CNOT, and Rz gates and observe the resulting state.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a two-qubit PennyLane device and QNode.
2. Apply Ry to the first qubit, CNOT between the qubits, and Rz to the second qubit.
3. Measure the final state and display the circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=2)

@qml.qnode(dev)
def circuit():
    qml.RY(np.pi/2,wires=0)
    qml.CNOT(wires=[0,1])
    qml.RZ(np.pi/2,wires=1)
    return qml.state()

state = circuit()

print("Final two-qubit state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The two-qubit circuit was successfully implemented. The Ry gate prepares a superposition, the CNOT introduces correlation between the qubits, and Rz changes the phase of the second qubit.
