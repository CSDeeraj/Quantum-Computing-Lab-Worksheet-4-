## Aim

To apply Rx followed by Ry on a single qubit and study the final quantum state.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a one-qubit PennyLane device and QNode.
2. Apply Rx(π/4) followed by Ry(π/3).
3. Measure the final state and display the circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)

@qml.qnode(dev)
def circuit():
    qml.RX(np.pi/4,wires=0)
    qml.RY(np.pi/3,wires=0)
    return qml.state()

state = circuit()

print("Final quantum state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The Rx and Ry gates were successfully applied sequentially. The final state demonstrates how successive rotations transform the qubit state.
