## Aim

To implement an Rx gate with rotation angle π/4 and observe the resulting quantum state.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Install PennyLane and open Jupyter Notebook or VS Code.
2. Create a one-qubit PennyLane device and QNode.
3. Apply Rx(π/4), measure the state, and display the circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)

@qml.qnode(dev)
def circuit():
    qml.RX(np.pi/4,wires=0)
    return qml.state()

state = circuit()

print("Quantum state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The Rx(π/4) gate was successfully applied to a single qubit. The resulting state vector and circuit confirm the correct implementation of the parameterized X-axis rotation.
