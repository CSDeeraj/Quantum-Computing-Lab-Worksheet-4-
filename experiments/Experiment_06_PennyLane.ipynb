## Aim

To compare the quantum states produced by Rx, Ry, and Rz gates when the same rotation angle π/4 is used.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a one-qubit PennyLane device.
2. Run separate QNodes using Rx, Ry, and Rz with angle π/4.
3. Display and compare their resulting states.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)
theta = np.pi/4

@qml.qnode(dev)
def rx_circuit():
    qml.RX(theta,wires=0)
    return qml.state()

@qml.qnode(dev)
def ry_circuit():
    qml.RY(theta,wires=0)
    return qml.state()

@qml.qnode(dev)
def rz_circuit():
    qml.RZ(theta,wires=0)
    return qml.state()

print("Rx(π/4):")
print(rx_circuit())
print("\nRy(π/4):")
print(ry_circuit())
print("\nRz(π/4):")
print(rz_circuit())

print("\nCircuits:")
print("Rx:",qml.draw(rx_circuit)())
print("Ry:",qml.draw(ry_circuit)())
print("Rz:",qml.draw(rz_circuit)())
```

## Result

For the same angle, Rx, Ry, and Rz produce different state transformations. Rx and Ry change amplitudes, while Rz primarily changes the relative phase.
