## Aim

To implement an Rz gate with rotation angle π/2 and observe its effect on the quantum state.

## Required Software

- Python 3.10 or above
- PennyLane
- Jupyter Notebook or Visual Studio Code
- NumPy


## Procedure

1. Create a one-qubit PennyLane device and QNode.
2. Apply Rz(π/2) to the qubit.
3. Measure the state and display the circuit.

## Program

The following program implements the experiment.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit",wires=1)

@qml.qnode(dev)
def circuit():
    qml.RZ(np.pi/2,wires=0)
    return qml.state()

state = circuit()

print("Quantum state:")
print(state)
print("\nCircuit:")
print(qml.draw(circuit)())
```

## Result

The Rz(π/2) gate was successfully implemented. The state vector shows the phase change produced by rotation about the Z-axis.
